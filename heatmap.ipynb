```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Data (replace this with your actual data)
data = {
    'Prompt': ['Prompt 1', 'Prompt 2', 'Prompt 3', 'Prompt 4', 'Prompt 5'],
    'No Emotional Input': [0.51, 0.46, 0.48, 0.79, 0.95],
    'Neutral': [0.44, 0.50, 0.49, 0.55, 0.46],
    'Happy': [0.63, 0.50, 0.54, 0.73, 0.50],
    'Sad': [0.62, 0.47, 0.50, 0.59, 0.73],
    'Angry': [0.64, 0.50, 0.46, 0.55, 0.63],
}

# Corresponding emotions for each data point
emotions = {
    'Prompt': ['Prompt 1', 'Prompt 2', 'Prompt 3', 'Prompt 4', 'Prompt 5'],
    'No Emotional Input': ['Optimistic', 'Approving', 'Sympathetic', 'Approving', 'Neutral'],
    'Neutral': ['Approving', 'Curious', 'Optimistic', 'Approving', 'Aware'],
    'Happy': ['Joyful', 'Joyful', 'Approving', 'Approving', 'Neutral'],
    'Sad': ['Sympathetic', 'Curious', 'Sympathetic', 'Sympathetic', 'Sympathetic'],
    'Angry': ['Approving', 'Angry', 'Approving', 'Sympathetic', 'Sympathetic'],
}

# Create DataFrames
df_values = pd.DataFrame(data).set_index('Prompt')
df_emotions = pd.DataFrame(emotions).set_index('Prompt')
```

```python
# Create a figure
fig, ax = plt.subplots(figsize=(10, 6))

# Create the heatmap with seaborn, set the colormap (e.g., Greens, Oranges) and range for values
sns.heatmap(df_values, annot=False, fmt="g", cmap='Purples', cbar=True, vmin=0, vmax=1, ax=ax)

for i in range(df_values.shape[0]):
    for j in range(df_values.shape[1]):
        emotion = df_emotions.iloc[i, j]     # Get the emotion label
        value = df_values.iloc[i, j]         # Get the numerical value
        # Combine emotion and number into one string
        text = f'{emotion}\n{value:.2f}'     # Format the number to 2 decimal places
        # Add the text to the heatmap
        ax.text(j + 0.5, i + 0.5, text, color='#240101', ha='center', va='center')

# Set labels and title
ax.set_title('Emotional Responses and Intensity Across Prompts and Emotional Inputs', fontsize=16)
plt.xlabel('Emotional Input')
plt.ylabel('Prompts')

# Show the plot
plt.tight_layout()
plt.show()
```
